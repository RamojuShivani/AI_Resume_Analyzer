```python
## first resume analyzer experiment

resume = """
I am a B.Tech graduate.
I know Python, SQL, Excel, and Machine Learning.
I have completed projects in data analysis.
"""

job_description = """
We are looking for a Python developer.
The candidate should know Python, SQL,
Machine Learning, and data analysis.
"""

print("RESUME:")
print(resume)

print("\nJOB DESCRIPTION:")
print(job_description)

## installing Scikit-learn
# Scikit-learn provides machine-learning tools, including the text-processing methods we'll use.

%pip install scikit-learn


from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [resume, job_description]

vectorizer = TfidfVectorizer(stop_words="english")

vectors = vectorizer.fit_transform(documents)

similarity = cosine_similarity(
    vectors[0:1],
    vectors[1:2]
)[0][0]

match_percentage = similarity * 100

print(f"Text similarity: {match_percentage:.2f}%")

# PyMuPDF? It's a Python library that allows us to read text from PDF files.

%pip install PyMuPDF


from google.colab import files

uploaded = files.upload()


```

```python

import fitz

# Get the uploaded PDF filename
pdf_filename = next(iter(uploaded))

# Open the PDF
document = fitz.open(pdf_filename)

# Extract text from every page
resume_text = ""

for page in document:
    resume_text += page.get_text() + "\n"

# Display the extracted text
print(resume_text)

document.close()
```

```python

# Skills our resume analyzer can recognize

skills_database = [
    "Python",
    "SQL",
    "Machine Learning",
    "Deep Learning",
    "Artificial Intelligence",
    "Power BI",
    "Tableau",
    "Excel",
    "Pandas",
    "NumPy",
    "Scikit-learn",
    "TensorFlow",
    "PyTorch",
    "Java",
    "C++",
    "HTML",
    "CSS",
    "JavaScript",
    "React",
    "FastAPI",
    "Flask",
    "Django",
    "AWS",
    "Google Cloud",
    "Docker",
    "Git",
    "GitHub",
    "Data Analysis",
    "Data Visualization",
    "Statistics"
]

print("Total skills in database:", len(skills_database))
```

```python

def extract_skills(text, skills_database):
    found_skills = []

    text_lower = text.lower()

    for skill in skills_database:
        if skill.lower() in text_lower:
            found_skills.append(skill)

    return found_skills


resume_skills = extract_skills(
    resume_text,
    skills_database
)

print("Skills found in your resume:")

for skill in resume_skills:
    print("-", skill)
```

```python

job_description = """
We are looking for a Python Developer.

Required skills:
Python, SQL, Machine Learning,
FastAPI, Docker, Git, AWS and Data Analysis.

The candidate should understand
problem-solving and software development.
"""

job_skills = extract_skills(
    job_description,
    skills_database
)

print("Skills required for the job:")

for skill in job_skills:
    print("-", skill)
```

```python

# Skills present in both the resume and job description
matched_skills = [
    skill for skill in job_skills
    if skill in resume_skills
]

# Required skills not found in the resume
missing_skills = [
    skill for skill in job_skills
    if skill not in resume_skills
]

print("MATCHING SKILLS:")

for skill in matched_skills:
    print("-", skill)

print("\nMISSING SKILLS:")

for skill in missing_skills:
    print("-", skill)
```

```python

if len(job_skills) > 0:
    match_percentage = (
        len(matched_skills) / len(job_skills)
    ) * 100
else:
    match_percentage = 0

print(f"Required skills found: {len(matched_skills)}")
print(f"Required skills: {len(job_skills)}")
print(f"Skill match: {match_percentage:.2f}%")
```

```python

def recommend_skills(missing_skills):
    recommendations = []

    learning_resources = {
        "Python": "Practice Python fundamentals and coding problems.",
        "SQL": "Learn joins, subqueries, and window functions.",
        "Machine Learning": "Study regression, classification, and model evaluation.",
        "FastAPI": "Build REST APIs using Python and FastAPI.",
        "Docker": "Learn containers, images, and Dockerfiles.",
        "AWS": "Study cloud fundamentals and deploy a small application.",
        "Google Cloud": "Learn cloud storage, computing, and deployment.",
        "Git": "Practice commits, branches, and merges.",
        "Power BI": "Build interactive dashboards using real datasets.",
        "Data Analysis": "Practice data cleaning, EDA, and visualization."
    }

    for skill in missing_skills:
        advice = learning_resources.get(
            skill,
            f"Learn the fundamentals of {skill} and build a small project."
        )

        recommendations.append({
            "skill": skill,
            "recommendation": advice
        })

    return recommendations


recommendations = recommend_skills(missing_skills)

for item in recommendations:
    print("Skill:", item["skill"])
    print("Action:", item["recommendation"])
    print("-" * 40)
```

```python

print("=" * 45)
print("AI RESUME ANALYZER")
print("=" * 45)

print(f"\nSkill Match: {match_percentage:.2f}%")

print("\nMATCHING SKILLS")
if matched_skills:
    for skill in matched_skills:
        print(f"  + {skill}")
else:
    print("  No matching skills found.")

print("\nSKILLS TO DEVELOP")
if missing_skills:
    for skill in missing_skills:
        print(f"  - {skill}")
else:
    print("  All recognized required skills were found.")

print("\nPERSONALIZED RECOMMENDATIONS")
if recommendations:
    for item in recommendations:
        print(f"  {item['skill']}: {item['recommendation']}")
else:
    print("  No additional skill recommendations.")

print("\nAnalysis completed.")
```

```python

%%writefile app.py

import streamlit as st
import fitz

# --------------------------------
# 1. Page configuration
# --------------------------------

st.set_page_config(
    page_title="AI Resume Analyzer",
    page_icon="📄",
    layout="wide"
)

# --------------------------------
# 2. Skills database
# --------------------------------

SKILLS = [
    "Python", "SQL", "Machine Learning",
    "Deep Learning", "Artificial Intelligence",
    "Power BI", "Tableau", "Excel",
    "Pandas", "NumPy", "Scikit-learn",
    "TensorFlow", "PyTorch", "Java", "C++",
    "HTML", "CSS", "JavaScript", "React",
    "FastAPI", "Flask", "Django", "AWS",
    "Google Cloud", "Docker", "Git", "GitHub",
    "Data Analysis", "Data Visualization",
    "Statistics", "NLP"
]

# --------------------------------
# 3. Extract text from PDF
# --------------------------------

def extract_pdf_text(pdf_file):
    text = ""

    with fitz.open(stream=pdf_file.read(),
                   filetype="pdf") as document:
        for page in document:
            text += page.get_text() + "\n"

    return text

# --------------------------------
# 4. Extract skills
# --------------------------------

def extract_skills(text):
    found = []
    text_lower = text.lower()

    for skill in SKILLS:
        if skill.lower() in text_lower:
            found.append(skill)

    return found

# --------------------------------
# 5. Learning recommendations
# --------------------------------

def get_recommendation(skill):
    recommendations = {
        "Python": "Practice Python fundamentals and coding.",
        "SQL": "Learn joins, subqueries and window functions.",
        "Machine Learning": "Study ML algorithms and evaluation.",
        "FastAPI": "Build REST APIs with Python.",
        "Docker": "Learn containers and Dockerfiles.",
        "AWS": "Study cloud basics and deployment.",
        "Git": "Practice commits, branches and merges.",
        "Power BI": "Build interactive business dashboards.",
        "Google Cloud": "Learn cloud computing and deployment."
    }

    return recommendations.get(
        skill,
        f"Learn {skill} fundamentals and build a small project."
    )

# --------------------------------
# 6. User interface
# --------------------------------

st.title("📄 AI-Powered Resume Analyzer")
st.write(
    "Upload your resume and compare your skills "
    "with a job description."
)

st.info(
    "This tool provides keyword-based skill matching. "
    "It does not predict hiring outcomes."
)

uploaded_file = st.file_uploader(
    "Upload your resume (PDF)",
    type=["pdf"]
)

job_description = st.text_area(
    "Paste the job description",
    height=220,
    placeholder="Paste the job requirements here..."
)

# --------------------------------
# 7. Analyze button
# --------------------------------

if st.button("Analyze Resume", type="primary"):

    if uploaded_file is None:
        st.error("Please upload a PDF resume.")

    elif not job_description.strip():
        st.error("Please enter a job description.")

    else:
        try:
            resume_text = extract_pdf_text(uploaded_file)

            if not resume_text.strip():
                st.error(
                    "No text found. This may be a scanned PDF."
                )

            else:
                resume_skills = extract_skills(resume_text)
                job_skills = extract_skills(job_description)

                matched = [
                    skill for skill in job_skills
                    if skill in resume_skills
                ]

                missing = [
                    skill for skill in job_skills
                    if skill not in resume_skills
                ]

                score = (
                    len(matched) / len(job_skills) * 100
                    if job_skills else 0
                )

                st.subheader("Analysis Results")

                col1, col2, col3 = st.columns(3)

                col1.metric(
                    "Skill Match",
                    f"{score:.1f}%"
                )

                col2.metric(
                    "Matching Skills",
                    len(matched)
                )

                col3.metric(
                    "Skills to Develop",
                    len(missing)
                )

                st.progress(score / 100)

                left, right = st.columns(2)

                with left:
                    st.subheader("✅ Matching Skills")

                    if matched:
                        for skill in matched:
                            st.write(f"✓ {skill}")
                    else:
                        st.write("No matching skills found.")

                with right:
                    st.subheader("🎯 Missing Skills")

                    if missing:
                        for skill in missing:
                            st.write(f"• {skill}")
                    else:
                        st.write(
                            "All recognized required skills "
                            "were found."
                        )

                st.subheader("📚 Learning Recommendations")

                if missing:
                    for skill in missing:
                        st.write(
                            f"**{skill}:** "
                            f"{get_recommendation(skill)}"
                        )
                else:
                    st.write(
                        "No additional recognized skills "
                        "to recommend."
                    )

                st.caption(
                    "A missing keyword does not necessarily "
                    "mean you lack that skill. Results depend "
                    "on the resume text and skills database."
                )

        except Exception as error:
            st.error(
                "Unable to analyze the PDF. "
                "Please check the file and try again."
            )
            st.exception(error)
```
